## Customizing Plots with Matplotlib

Below are code examples for customizing plots using Matplotlib, including legends, annotations, axis labels, and styles. 

```python
# Importing libraries for matplotlib and numpy
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
# Set random seed and create sample data for plotting
data_rng = np.random.default_rng(613)
x = np.arange(50)
y1 = data_rng.integers(0, 100, 50)
y2 = data_rng.integers(0, 100, 50)
```

```python
# Simple line plot with two y variables
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y1)
ax.plot(x, y2)
fig.show()  # Show both y variables on the same axes
```

```python
# Add legend to the plot
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y1, label="Person 1")
ax.plot(x, y2, label="Person 2")
ax.legend(loc='lower right')
fig.show()  # Adds a legend to distinguish lines
```

```python
# Customizing the legend (frame, font size, columns, shadow)
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y1, label="Person 1")
ax.plot(x, y2, label="Person 2")
ax.legend(loc='lower right', frameon=True, fontsize=12, ncol=2, shadow=True)
fig.show()  # Customizes legend appearance
```

```python
# Move legend outside the plot area
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y1, label="Person 1")
ax.plot(x, y2, label="Person 2")
ax.legend(loc='upper left', bbox_to_anchor=(1, 1))
fig.show()  # Legend anchored outside plot
```

```python
# Scatter plot with text annotation
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x, y1, label="Person 1")
ax.scatter(x, y2, label="Person 2")
ax.legend(loc='lower right')
ax.text(10, 95, "This value is important!")
fig.show()  # Adds a text annotation
```

```python
# Modify text annotation (alignment, color, size)
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x, y1, label="Person 1")
ax.scatter(x, y2, label="Person 2")
ax.legend(loc='lower right')
ax.text(10, 95, "This value is important!", ha='center', color='red', size=20)
fig.show()  # Customizes annotation style
```

```python
# Positioning text annotations using transforms
fig, ax = plt.subplots()
ax.axis([0, 10, 0, 10])
ax.text(1, 5, ". Data: (1, 5)", transform=ax.transData)
ax.text(0.5, 0.1, ". Axes: (0.5, 0.1)", transform=ax.transAxes)
ax.text(0.2, 0.2, ". Figure: (0.2, 0.2)", transform=fig.transFigure)
fig.show()  # Shows annotation at different reference points
```

```python
# Annotate with arrows
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x, y1, label="Person 1")
ax.scatter(x, y2, label="Person 2")
ax.legend(loc='lower right')
ax.annotate('This is important!', xy=(10, 95), xytext=(20, 94), arrowprops=dict(facecolor='black'))
fig.show()  # Adds an arrow annotation
```

```python
# Custom arrow style and color
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x, y1, label="Person 1")
ax.scatter(x, y2, label="Person 2")
ax.legend(loc='lower right')
ax.annotate('This is important!', xy=(10, 95), xytext=(20, 94), arrowprops=dict(arrowstyle="wedge", color="hotpink"))
fig.show()  # Customizes arrow style
```

```python
# Remove tick marks and labels
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x, y1, label="Person 1")
ax.scatter(x, y2, label="Person 2")
ax.legend(loc='lower right')
ax.yaxis.set_major_locator(plt.NullLocator())
ax.xaxis.set_major_formatter(plt.NullFormatter())
fig.show()  # Removes axis ticks and labels
```

```python
# Limit number of x-axis ticks
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x, y1, label="Person 1")
ax.scatter(x, y2, label="Person 2")
ax.legend(loc='lower right')
ax.xaxis.set_major_locator(plt.MaxNLocator(3))
fig.show()  # Limits number of ticks
```

```python
# Set tick marks at interval of 5
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x, y1, label="Person 1")
ax.scatter(x, y2, label="Person 2")
ax.legend(loc='lower right')
ax.xaxis.set_major_locator(plt.MultipleLocator(5))
fig.show()  # Sets tick interval
```

```python
# Rotate x-axis labels
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x, y1, label="Person 1")
ax.scatter(x, y2, label="Person 2")
ax.legend(loc='lower right')
plt.xticks(rotation=45, ha='right')
fig.show()  # Rotates x-axis labels
```

```python
# Modify x-axis title font and color
font1 = {'family': 'serif', 'color': 'indigo'}
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x, y1, label="Person 1")
ax.scatter(x, y2, label="Person 2")
ax.legend(loc='lower right')
plt.xlabel('Shiny New X Axis!', fontsize=18, fontdict=font1)
fig.show()  # Customizes x-axis title
```

```python
# Use a pre-made style for plots
plt.style.use('fivethirtyeight')
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y1)
ax.plot(x, y2)
fig.show()  # Applies a style to the plot
```


## Beyond Matplotlib: Seaborn, Plotly, Wordclouds, Venn Diagrams

Below are code examples for using Seaborn, Plotly, wordclouds, and Venn diagrams for advanced and interactive data visualization. 


```python
# Seaborn: import and load tips dataset
import seaborn as sns
tips = sns.load_dataset("tips")
print(tips.head())  # Loads sample dataset for visualization
```

```python
# Simple Seaborn line plot
tipgraph = sns.lineplot(data=tips, x='total_bill', y='tip')
plt.show()  # Line plot of tip vs total bill
```

```python
# Use a premade Seaborn style
sns.set_style('whitegrid')
tipgraph = sns.lineplot(data=tips, x='total_bill', y='tip')
plt.show()  # Applies 'whitegrid' style
```

```python
# Add title and axis labels in Seaborn
tipgraph = sns.lineplot(data=tips, x='total_bill', y='tip')
tipgraph.set(title='Tips vs. Total Bill', xlabel='Total Bill ($)', ylabel='Tip Amount ($)')
plt.show()  # Adds title and axis labels
```

```python
# Change figure size with matplotlib
fig, _ = plt.subplots(figsize=(10, 3))
plt.show()  # Changes plot size
```

```python
# Change color, marker, and line style in Seaborn
tipgraph = sns.lineplot(data=tips, x='total_bill', y='tip', color='hotpink', linestyle='--', linewidth=3, marker='o', markerfacecolor='indigo')
plt.show()  # Customizes line and markers
```

```python
# Scatterplot with multiple variables (style and color by category)
tipgraph = sns.scatterplot(data=tips, x='total_bill', y='tip', style='time', hue='day', palette=['purple', 'hotpink', 'deepskyblue', 'yellowgreen'])
tipgraph.set(title='Tips vs. Total Bill', xlabel='Total Bill ($)', ylabel='Tip Amount ($)')
plt.show()  # Encodes more variables visually
```

```python
# Pairplot to compare variables
sns.pairplot(data=tips, hue='day')
plt.show()  # Shows pairwise relationships
```

```python
# Relplot to explore levels within variables
daysplot = sns.relplot(data=tips, x="total_bill", y="tip", hue="sex", col="day", kind="scatter", col_wrap=2)
plt.show()  # Faceted scatter plots by day and sex
```

```python
# Plotly: bar plot example
import plotly.graph_objects as go
x1 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y1 = np.array([110, 180, 240, 99, 220])
graph = go.Figure()
graph.add_trace(go.Bar(x=x1, y=y1))
graph.update_layout(title="Pirate Scores", xaxis_title="Pirates", yaxis_title="Score")
graph.show()  # Interactive bar plot
```

```python
# Plotly: scatter plot with custom markers
graph = go.Figure()
graph.add_trace(go.Scatter(x=x1, y=y1, mode='markers', marker=dict(size=15, color='hotpink', opacity=1, line=dict(width=5, color='purple'))))
graph.update_layout(title='Interactive Pirate Plot', xaxis_title='Pirates', yaxis_title='Scores', width=500, height=500)
graph.show()  # Custom interactive scatter plot
```

```python
# Wordcloud: generate from text data
from wordcloud import WordCloud
df = pd.read_csv("https://raw.githubusercontent.com/prasertcbs/basic-dataset/master/movie_quotes.csv", on_bad_lines='skip')
text = " ".join(each for each in df.quote)
wordcloud = WordCloud(background_color="white", colormap='inferno').generate(text)
fig, ax = plt.subplots(figsize=(7, 3))
ax.imshow(wordcloud, interpolation='bilinear')
ax.axis("off")
plt.show()  # Generates a word cloud from quotes
```

```python
# Venn diagram with matplotlib_venn
from matplotlib_venn import venn2, venn2_circles, venn2_unweighted
A = set(["apple", "banana", "watermelon"])
B = set(["pumpkin", "blueberry", "apple", "key lime"])
diagram = venn2_unweighted([A, B], set_labels=('Fruits', 'Pies'), set_colors=("blue", "red"), alpha=0.5)
plt.show()  # Simple Venn diagram
```

```python
# Customizing Venn diagram labels
diagram.get_label_by_id("10").set_text("\n".join(A - B))
diagram.get_label_by_id("11").set_text("\n".join(A & B))
diagram.get_label_by_id("01").set_text("\n".join(B - A))
plt.show()  # Customizes Venn diagram region labels
```
