```python
# Import core libraries for visualization
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Set random seed for consistent results
np.random.seed(120334)
x = np.arange(50)
y = np.random.randint(0, 100, 50)
print(f"Generated {len(x)} data points")
```

```python
# Simple scatter plot example
fig, ax = plt.subplots(figsize=(6, 4))
ax.scatter(x, y, alpha=0.7)
plt.show()
```

```python
# Bar chart visualization
fig, ax = plt.subplots(figsize=(6, 4))
ax.bar(x, y, color='steelblue')
plt.show()
```

```python
# Line plot visualization
fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(x, y, color='teal')
plt.show()
```

```python
# Line plot with axis labels and title
fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(x, y, color='darkgreen')
ax.set_title('Growth Trend Over Time')
ax.set_ylabel('Total Growth')
ax.set_xlabel('Time (Years)')
fig.tight_layout()
plt.show()
```

```python
# Styled line plot with custom markers
fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(x, y,
        marker='o',
        markersize=8,
        color='#5a2d82',
        linestyle='-.',
        linewidth=1.5,
        markeredgecolor='orange',
        markerfacecolor='orange')
plt.show()
```

```python
# Generate data for 2D density visualization
size = 80000
df = pd.DataFrame({
    'x': np.random.normal(loc=0, scale=1, size=size),
    'y': np.random.normal(loc=0, scale=1, size=size)
})
df.head(10)
```

```python
# Hexbin density plot
fig, ax = plt.subplots(figsize=(8, 6))
hb = ax.hexbin(df['x'], df['y'], gridsize=25, cmap='Purples')
ax.set_xlabel('X Axis')
ax.set_ylabel('Y Axis')
ax.set_title('Hexagonal Binning Density')
cb = fig.colorbar(hb, ax=ax)
cb.set_label('Frequency')
plt.show()
```

```python
# Line plot with horizontal grid
fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(x, y, color='coral')
ax.grid(axis='y', color='gray', linewidth=0.4, linestyle='--')
ax.set_title('Plot with Gridlines')
plt.show()
```

```python
# Histogram of y values
fig, ax = plt.subplots(figsize=(6, 4))
ax.hist(y, bins=12, edgecolor='white', color='mediumpurple')
ax.set_xlabel('Value Range')
ax.set_ylabel('Count')
ax.set_title('Y Value Distribution')
plt.show()
```
